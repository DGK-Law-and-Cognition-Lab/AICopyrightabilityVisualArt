# Study 2, part 2: measuring what participants wrote

Each of the 235 participants rated second works on transformativeness, essence change and creativity, and
wrote a justification after every rating. Registration question 6 drops the two six-point
transformativeness items with their justifications, which leaves 20,659 justifications. This notebook
turns them into numbers. It applies Study 1's eight word lists, the rules, without changing them, and
writes the tables the integration file (`Study2_03_Integration.Rmd`) needs for the two registered text
checks and for the exploratory description.

**The measurement is blind by construction.** The notebook reads four columns of the long data file: the
participant, the second work, the question and the text. It never reads the source condition, the
modification label the models use or the rating, so nothing here can depend on them, and it has no blind
and unblinded modes. The integration file joins the condition and the rating back on.

No person and no language model labels, codes or judges any response. A rule is a search pattern (a
written list of words and phrases, in any of their ordinary forms) that the computer applies in the same
way to every response. The one use of a pretrained model is the check in section 5, in which a locally
stored sentence-embedding model (all-mpnet-base-v2) turns each response into a list of numbers so that a
grouping by meaning can be compared with the topics.

Everything is computed by the shared module `shared/python/study_text.py`, which holds the measurement
half of the Study 1 text notebook. This notebook chooses the input, checks it, and names the files.

Run it from this folder with the Python 3.12 kernel:

    HF_HUB_OFFLINE=1 TRANSFORMERS_OFFLINE=1 jupyter nbconvert --to notebook --execute --inplace \
      --ExecutePreprocessor.kernel_name=python312 Study2_02_Text.ipynb

# 0. Setup

The paths start from this notebook's folder. Three environment variables (settings the shell passes to the
program) can name other folders: `STUDY2_DIR` the Study 2 folder, `STUDY1_DIR` the Study 1 folder whose
deposited rules are compared below, and `STUDY_SHARED_PYTHON` the folder of the shared module. Printed
paths are relative to the pipeline folder, so the output names no account and no folder particular to one
machine.

The input is `study2_long_with_text.csv`, which `Study2_01_Quant.Rmd` writes to the Study 2 folder in an
unblinded run and to `blind_run/` in a blind run. The two files hold the same participants, second works,
questions and text in the same order, because a blind run shuffles only labels this notebook does not
read. The notebook reads the unblinded file when it exists and the blind file otherwise, and says which.

The registration says that resampling uses the Study 1 seed, 20260421. A seed is the number that starts a
random-number generator. The stability rule of the topic model refits the model on random subsets of the
participants, which is resampling, so the cell below makes 20260421 the seed of every random step in this
notebook, all of which sit in the topic model and its agreement check. The seed alone does not make a
rerun give the same numbers. The shared module also runs the k-means step of the agreement check, which
splits the responses into groups of similar meaning, on one processor thread (one strand of computation).
On several threads, k-means adds up its numbers in pieces, one per thread, and combines the pieces in the
order the threads finish, which changes from run to run; since rounding depends on that order, its groups
can differ. The seed and the single thread together make a rerun give the same numbers.

```python
import datetime
import hashlib
import os
import platform
import sys
import warnings

# The Hugging Face libraries that load the embedding model are kept off the
# internet, so nothing is downloaded.
os.environ.setdefault("HF_HUB_OFFLINE", "1")
os.environ.setdefault("TRANSFORMERS_OFFLINE", "1")


def short_warning(message, category, filename, lineno, line=None):
    # A warning names the file it came from without the file's folder.
    return "%s: %s (%s, line %d)\n" % (category.__name__, message, os.path.basename(filename), lineno)


warnings.formatwarning = short_warning

import numpy as np
import pandas as pd

STUDY_DIR = os.path.abspath(os.environ.get("STUDY2_DIR", os.getcwd()))
STUDY1_DIR = os.path.abspath(os.environ.get("STUDY1_DIR", os.path.join(STUDY_DIR, os.pardir, "Study 1")))
SHARED_PYTHON = os.path.abspath(os.environ.get("STUDY_SHARED_PYTHON",
                                               os.path.join(STUDY_DIR, os.pardir, "shared", "python")))
sys.dont_write_bytecode = True
sys.path.insert(0, SHARED_PYTHON)
import study_text

# The registered seed. The ten refits of the topic model's stability rule use
# 20260421 to 20260430, and the final topic fit and the k-means grouping of
# the agreement check use 20260421.
TEXT_SEED = 20260421
TEXT_SEEDS = study_text.set_text_seed(TEXT_SEED)

show = study_text.display_path
INPUT_NAME = "study2_long_with_text.csv"
CANDIDATES = [os.path.join(STUDY_DIR, INPUT_NAME), os.path.join(STUDY_DIR, "blind_run", INPUT_NAME)]
INPUT_PATH = next((p for p in CANDIDATES if os.path.exists(p)), None)
assert INPUT_PATH is not None, (
    "Cannot find %s or %s. Knit Study2_01_Quant.Rmd first." % tuple(show(p) for p in CANDIDATES))

NLP_DIR = os.path.join(STUDY_DIR, "nlp")
TABLES_DIR = os.path.join(STUDY_DIR, "tables")
os.makedirs(NLP_DIR, exist_ok=True)
os.makedirs(TABLES_DIR, exist_ok=True)

# The previous run's manifest is deleted before this run writes anything, so
# a manifest on disk means the run that wrote it reached the last cell.
MANIFEST_PATH = os.path.join(TABLES_DIR, "study2_text_run_manifest.csv")
if os.path.exists(MANIFEST_PATH):
    os.remove(MANIFEST_PATH)

RUN_ID = study_text.new_run_id()


def file_sha256(path):
    # The SHA-256 checksum of a file's bytes: a 64-character string that
    # changes if any byte of the file changes.
    digest = hashlib.sha256()
    with open(path, "rb") as handle:
        for block in iter(lambda: handle.read(1 << 20), b""):
            digest.update(block)
    return digest.hexdigest()


INPUT_SHA256 = file_sha256(INPUT_PATH)
print("Input:          ", show(INPUT_PATH))
print("Input SHA-256:  ", INPUT_SHA256)
print("Writing data to:", show(NLP_DIR))
print("Writing tables: ", show(TABLES_DIR))
print("Run identifier: ", RUN_ID)
print("Seeds:          ", TEXT_SEEDS)
```

# 1. The blind slice

The blind slice is the only table the measurement sees. Only four columns are loaded from the long data
file, so the condition, the modification label the models use and the rating never enter this
notebook's memory.

A response is about one second work: a source image at one modification level, as it was shown. The key
column `stimulus_id` therefore names the second work, such as `3NonFigure_Slight`, and the source image,
the image kind and the modification level shown travel beside it as `stimulus_image`, `stimulus_type` and
`stimulus_modification`. All three are read off the second work's name, which a blind run never shuffles.
The shared module accepts only columns named `stimulus_...` beside the key and the text, and refuses any
other column by name.

```python
READ_COLUMNS = ["id", "work_id", "dimension", "response_text"]
raw = pd.read_csv(INPUT_PATH, usecols=READ_COLUMNS, dtype=str, keep_default_na=False)
assert sorted(raw.columns) == sorted(READ_COLUMNS), "Unexpected columns were read."

parts = raw["work_id"].str.extract(r"^((\d)(Figure|NonFigure))_(None|Slight|Dramatic)$")
assert not parts.isna().any().any(), "Some second work is not named image_level."
blind = pd.DataFrame({
    "id": raw["id"],
    "stimulus_id": raw["work_id"],
    "stimulus_image": parts[0],
    "stimulus_type": parts[2],
    "stimulus_modification": parts[3],
    "dimension": raw["dimension"],
    "response_text": raw["response_text"],
})
del raw, parts
study_text.check_blind(blind)

# The counts the registration states: 20,659 justifications from 235
# participants once the two six-point items are dropped, none empty. A
# participant who saw all 30 second works wrote 88 of them, and the seven who
# lack the trial 1FigureMin wrote 85.
n_rows, n_people = len(blind), blind["id"].nunique()
per_person = blind.groupby("id").size()
print("Responses:", n_rows, "| participants:", n_people,
      "| second works:", blind["stimulus_id"].nunique(),
      "| source images:", blind["stimulus_image"].nunique())
print("Questions:", sorted(blind["dimension"].unique()))
print("Responses per participant:", per_person.value_counts().sort_index().to_dict())
assert n_rows == 20659 and n_people == 235
assert blind["stimulus_id"].nunique() == 30 and blind["stimulus_image"].nunique() == 10
assert per_person.value_counts().to_dict() == {88: 228, 85: 7}
assert (blind["response_text"].str.strip() != "").all(), "A justification is empty."
print("Columns:", list(blind.columns))
```

# 2. The rules are Study 1's

The registration says Study 1's eight word lists run unchanged. Two checks show that they do.

First, the module's rules, written out as Study 1's notebook wrote them (with Study 1's name and its
2,220 responses), must equal Study 1's deposited `tables/text_rules.txt` character for character, and
their SHA-256 checksum must equal the one in Study 1's `tables/text_rules.sha256`.

Second, Study 2's own rules text may differ from Study 1's in two lines only: the title, which names the
study, and the line that gives the number of responses. The constructed sentences, written for the check
and taken from no corpus, must also be read as the rules say.

```python
STUDY1_RULES = os.path.join(STUDY1_DIR, "tables", "text_rules.txt")
STUDY1_RULES_SHA = os.path.join(STUDY1_DIR, "tables", "text_rules.sha256")
STUDY1_N_RESPONSES = 2220

with open(STUDY1_RULES, encoding="utf-8") as handle:
    study1_rules_text = handle.read()
with open(STUDY1_RULES_SHA, encoding="utf-8") as handle:
    study1_sha = handle.read().split()[0]

as_study1 = study_text.format_rules(study_text.RULES, "Study 1", STUDY1_N_RESPONSES) + "\n"
assert as_study1 == study1_rules_text, "The module's rules differ from Study 1's deposited rules."
assert study_text.sha256_text(as_study1) == study1_sha == study_text.sha256_text(study1_rules_text), \
    "The SHA-256 of the rules differs from Study 1's deposited checksum."
print("Rules equal Study 1's tables/text_rules.txt: True")
print("SHA-256 equals Study 1's tables/text_rules.sha256:", study1_sha)

RULES_TEXT = study_text.rules_text(blind, "Study 2")
RULES_SHA256 = study_text.sha256_text(RULES_TEXT)
assert len(study1_rules_text.splitlines()) == len(RULES_TEXT.splitlines())
differing = [(a, b) for a, b in zip(study1_rules_text.splitlines(), RULES_TEXT.splitlines()) if a != b]
assert len(differing) == 2, "Study 2's rules differ from Study 1's in more than the two named lines."
print()
print("Lines in which Study 2's rules text differs from Study 1's:")
for a, b in differing:
    print("  Study 1:", a)
    print("  Study 2:", b)
print()
print("SHA-256 of Study 2's rules text:", RULES_SHA256)

probes = study_text.check_constructed_sentences()
print("Constructed sentences read as the rules say:", len(probes), "of", len(probes))
```

# 4. The word tables for the two registered checks

Three tables carry the words of each response to the integration file. The word-count table has one row
per response and content word (a word left after the fixed list of function words such as *the* and *of*
is removed) that appears in at least two responses; the vocabulary lists those words with the number of
responses each appears in; the content-word table gives each response's distinct content words. The first
check predicts the rating and the source condition from the counts. The second compares how much
participants reuse their own words, using the content-word lists. The registration fixes both checks for
creativity and names the same steps on the other two questions as exploratory, so the tables cover every
question.

```python
WORDCOUNTS_PATH = os.path.join(NLP_DIR, "study2_text_wordcounts.csv")
VOCABULARY_PATH = os.path.join(NLP_DIR, "study2_text_vocabulary.csv")
TOKENS_PATH = os.path.join(NLP_DIR, "study2_text_tokens.csv")
wordcounts = study_text.wordcounts_frame(blind)
vocabulary = study_text.vocabulary_frame(blind)
tokens = study_text.tokens_frame(blind)
study_text.write_csv_atomic(WORDCOUNTS_PATH, wordcounts, RUN_ID)
study_text.write_csv_atomic(VOCABULARY_PATH, vocabulary, RUN_ID)
study_text.write_csv_atomic(TOKENS_PATH, tokens, RUN_ID)
for path, frame in [(WORDCOUNTS_PATH, wordcounts), (VOCABULARY_PATH, vocabulary), (TOKENS_PATH, tokens)]:
    print("Written:", show(path), "|", len(frame), "rows")
```

# 5. The exploratory description

**Tone and concreteness.** Each response gets the mean valence (how pleasant a word is), arousal (how
excited it makes a reader feel), dominance (how much in control it makes a reader feel) and concreteness
of its content words, from published word ratings (Warriner, Kuperman & Brysbaert, 2013; Brysbaert,
Warriner & Kuperman, 2014). The words of the rating questions are removed first. A response with no
matched word gets an empty score.

**Topics.** A topic model finds groups of words that tend to occur in the same responses. The model is
non-negative matrix factorisation (it writes the table of responses by words as the product of two tables
with no negative entries, one giving each response's weight on each topic), fitted to each question
separately and given no labels, with the number of topics chosen by the stability rule in the module.

**The agreement check.** The sentence-embedding model turns each response into 768 numbers, the responses
to each question are split into as many groups as that question has topics, and the split is compared with
each response's main topic by the adjusted Rand index (agreement between two groupings beyond chance: 0
for chance, 1 for identical). The ten refits of the stability rule use the seeds 20260421 to 20260430, and
the final topic fit and the grouping use 20260421. Nothing the embedding model produces is kept beyond
that index.

```python
AFFECT_PATH = os.path.join(NLP_DIR, "study2_text_affect.csv")
affect = study_text.affect_frame(blind)
study_text.write_csv_atomic(AFFECT_PATH, affect, RUN_ID)
print("Written:", show(AFFECT_PATH), "|", len(affect), "rows")
print("Responses with no word in the valence norms:", int((affect["n_vad_matched"] == 0).sum()))
print("Responses with no word in the concreteness norms:", int((affect["n_conc_matched"] == 0).sum()))
```

```python
TOPICS_PATH = os.path.join(NLP_DIR, "study2_text_topics.csv")
TOP_WORDS_PATH = os.path.join(TABLES_DIR, "study2_text_topics_top_words.csv")
STABILITY_PATH = os.path.join(TABLES_DIR, "study2_text_topics_stability.csv")
AGREEMENT_PATH = os.path.join(TABLES_DIR, "study2_text_topics_embedding_agreement.csv")

topics = study_text.topics_frame(blind)
top_words = study_text.topic_top_words_frame(blind)
stability = study_text.topic_stability_frame(blind)
fit = study_text.fit_topics(blind)
study_text.write_csv_atomic(TOPICS_PATH, topics, RUN_ID)
study_text.write_csv_atomic(TOP_WORDS_PATH, top_words, RUN_ID)
study_text.write_csv_atomic(STABILITY_PATH, stability, RUN_ID)
for dim, fitted in fit["by_dimension"].items():
    print("%-18s topics: %d | vocabulary: %d | refits stopped at the iteration limit: %d of %d | final fit settled: %s"
          % (dim, fitted["k"], fitted["n_vocab"], fitted["n_short"], fitted["n_refits"],
             fitted["final_converged"]))
print()
print(top_words[top_words["rank"] <= 5].groupby(["dimension", "topic"])["word"]
      .apply(", ".join).to_string())

agreement = study_text.embedding_agreement_frame(blind)
EMBEDDING_STATUS = agreement.attrs.get("embedding_status", "")
study_text.write_csv_atomic(AGREEMENT_PATH, agreement, RUN_ID)
print()
print("Embedding model:", study_text.EMBEDDING_MODEL_NAME, "|", EMBEDDING_STATUS)
print(agreement.to_string(index=False))
```

# 6. Export

The rules, their checksum and their grounding notes go to `tables/`, with a record of the software and a
run manifest. Every file name carries the `study2_` prefix, and the checksum line names
`study2_text_rules.txt`, so `shasum -a 256 -c study2_text_rules.sha256` run from `tables/` checks it.
Every data file is read back from disk and held to its agreed columns, none of which is the condition, the
rating or the text. The manifest lists every output with its size and SHA-256, beside the run identifier,
the input's SHA-256 and the rules checksum, and it is written last.

```python
RULES_PATH = os.path.join(TABLES_DIR, "study2_text_rules.txt")
RULES_SHA_PATH = os.path.join(TABLES_DIR, "study2_text_rules.sha256")
GROUNDING_PATH = os.path.join(TABLES_DIR, "study2_text_rules_grounding.txt")
ENVIRONMENT_PATH = os.path.join(TABLES_DIR, "study2_text_environment.csv")

study_text.write_text_atomic(RULES_PATH, RULES_TEXT, RUN_ID)
study_text.write_text_atomic(RULES_SHA_PATH, "%s  study2_text_rules.txt\n" % RULES_SHA256, RUN_ID)
study_text.write_text_atomic(GROUNDING_PATH, study_text.grounding_text(blind, "Study 2"), RUN_ID)
assert file_sha256(RULES_PATH) == RULES_SHA256


def package_version(module_name):
    try:
        module = __import__(module_name)
    except Exception:
        return "not installed"
    return str(getattr(module, "__version__", "installed, version not recorded"))


environment = pd.DataFrame([
    ("run_id", RUN_ID),
    ("python", platform.python_version()),
    ("platform", platform.platform()),
    ("pandas", pd.__version__),
    ("numpy", np.__version__),
    ("scikit-learn", package_version("sklearn")),
    ("scipy", package_version("scipy")),
    ("sentence-transformers", package_version("sentence_transformers")),
    ("torch", package_version("torch")),
    ("embedding_model", "%s, %s" % (study_text.EMBEDDING_MODEL_NAME, EMBEDDING_STATUS)),
    ("seed_text_code", str(TEXT_SEED)),
    ("seed_stability_refits", "%d to %d" % (TEXT_SEEDS["stability_refits_first"],
                                            TEXT_SEEDS["stability_refits_last"])),
    ("seed_final_topic_fit", str(TEXT_SEEDS["final_topic_fit"])),
    ("seed_kmeans_grouping", str(TEXT_SEEDS["kmeans_grouping"])),
    ("kmeans_openmp_threads", str(study_text.KMEANS_OPENMP_THREADS)),
], columns=["item", "value"])
# The seeds in force at the end of the run are the ones set at its start.
assert study_text.text_seeds() == TEXT_SEEDS
study_text.write_csv_atomic(ENVIRONMENT_PATH, environment, RUN_ID)
print(environment.to_string(index=False))
print()

# Every data file, read back from disk and held to its agreed columns. Every
# file keyed by response, apart from the word-count table (which has no row
# for a response with no counted word), covers exactly the 20,659 responses.
FORBIDDEN = ["Type", "source", "modification", "image_kind", "rating", "response_text",
             "condition", "arm", "age", "gender", "experience"]
DATA_CONTRACTS = [
    (MEASURES_PATH, study_text.export_columns(blind)),
    (WORDCOUNTS_PATH, study_text.WORDCOUNT_COLUMNS),
    (VOCABULARY_PATH, study_text.VOCABULARY_COLUMNS),
    (TOKENS_PATH, study_text.TOKENS_COLUMNS),
    (AFFECT_PATH, study_text.AFFECT_COLUMNS),
    (TOPICS_PATH, study_text.TOPICS_COLUMNS),
]
response_keys = set(blind[study_text.KEY].itertuples(index=False, name=None))
for path, columns in DATA_CONTRACTS:
    on_disk = pd.read_csv(path, dtype=str, keep_default_na=False)
    assert list(on_disk.columns) == list(columns), "%s has the wrong columns." % show(path)
    study_text.check_output(on_disk, show(path))
    assert not [c for c in FORBIDDEN if c in on_disk.columns], "%s carries a forbidden column." % show(path)
    if path not in (WORDCOUNTS_PATH, VOCABULARY_PATH):
        keys = set(on_disk[study_text.KEY].itertuples(index=False, name=None))
        assert keys == response_keys, "%s does not cover exactly the 20,659 responses." % show(path)
    print("  %-38s %7d rows, %2d columns as agreed, no forbidden column"
          % (show(path), len(on_disk), len(columns)))

outputs = [MEASURES_PATH, WORDCOUNTS_PATH, VOCABULARY_PATH, TOKENS_PATH, AFFECT_PATH, TOPICS_PATH,
           TOP_WORDS_PATH, STABILITY_PATH, AGREEMENT_PATH, RULES_PATH, RULES_SHA_PATH,
           GROUNDING_PATH, ENVIRONMENT_PATH]
completed_utc = datetime.datetime.now(datetime.timezone.utc).strftime("%Y-%m-%dT%H:%M:%SZ")
manifest = pd.DataFrame([dict(
    run_id=RUN_ID, completed_utc=completed_utc,
    input_file=os.path.relpath(INPUT_PATH, STUDY_DIR), input_sha256=INPUT_SHA256,
    rules_sha256=RULES_SHA256,
    file=os.path.relpath(path, STUDY_DIR), bytes=os.path.getsize(path), sha256=file_sha256(path),
) for path in outputs])
study_text.write_csv_atomic(MANIFEST_PATH, manifest, RUN_ID)
print()
print("Manifest:", show(MANIFEST_PATH), "|", len(manifest), "outputs listed, written last.")
```

## What this notebook does not write

Study 1's text notebook also writes four reports that this notebook does not: the matched and unmatched
responses for each rule and the list of candidate words, which are printed for a person to read, and the
distinctive-word tables and the corpus summary by arm, which need the rating or the condition. The
measurement here never reads either.
